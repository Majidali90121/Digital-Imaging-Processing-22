## If , elif and else

```python
testValue=[-1,0,127,128,255,256]
for intensity in testValue:
  if intensity<0 or intensity>255:
    print("invalid")
  elif intensity>=128:
    print("bright")
  else:
    print("Dark")
```

## Using For and While loop

```python
value=[20,130,255]

for pixel in value:
    if pixel>=128:
        print(f"{pixel}: Bright pixel")
    else:
        print(f"{pixel}: Dark Pixel")


#while loop

a=0

while a<3:
    print(f"value of {a}")
    a+=1
```

## Using Fuction

```python
def is_bright(pixel,thresthold=128):
    if pixel>=thresthold:
        return True
    else:
        return False

print(is_bright(127))
print(is_bright(128))
print(is_bright(200,210))
```
