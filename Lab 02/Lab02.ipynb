## Importing Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Reading Images

```python
img=cv2.imread("image.jpg")

print(f"Image Shape: {img.shape}")
print(f"Image Data Type: {img.dtype}")
```

## Inspect Image

```python
import numpy as np
h,w,c=img.shape
print(f"Height: {h} Width: {w} and Channel: {c}")

#top left pixel 0 rows 0 column
top_left_pixel=img[0,0]
#center Pixel by tasking by integer devision to find exact middle h//2 w//2
center_pixel=img[h//2,w//2]

print(f"top_left_pixel: {top_left_pixel} center Pixel: {center_pixel}")

#Min max funtions

global_min=np.min(img)
global_max=np.max(img)

print(f"Global Min: {global_min} and Global Max: {global_max}")

#speration of channel (B,G,R) based on 3D pixels
b,g,r=cv2.split(img)

print(f"Blue- Min: {np.min(b)} Max:{np.max(b)}")
print(f"Green- Min: {np.min(g)} Max:{np.max(g)}")
print(f"Red- Min: {np.min(r)} Max:{np.max(r)}")
```

## Converte Colors And Seprate Channels

```python
import matplotlib.pyplot as plt
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

#gray Scale Version
img_gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

#seprate image using channels

r,g,b=cv2.split(img)

#4show 2X3 figure

fig,ax=plt.subplots(2,3,figsize=(15,10))

ax[0,0].imshow(img_rgb); ax[0,0].set_title("RGB Image")
ax[0,1].imshow(r,cmap='gray'); ax[0,1].set_title("Red Channel")
ax[0,2].imshow(g,cmap='gray'); ax[0,2].set_title("Green Channel")
ax[1,0].imshow(b,cmap='gray'); ax[1,0].set_title("Blue Channel")
ax[1,1].imshow(img_gray,cmap='gray'); ax[1,1].set_title("GrayScale Image")
ax[1,2].imshow(img); ax[1,2].set_title("BGR Image (No conversion)")

#5 hidden pixels and show image

plt.axis("off")
plt.tight_layout()
plt.show()

#6 Center pixels
center_rgb=img_rgb[h//2,w//2]
hex_color='#{:02x}{:02x}{:02x}'.format(center_rgb[0],center_rgb[1],center_rgb[2])
print(f"Center Pixel RGB: {center_rgb} | HEX: {hex_color}")
```

## Crop Resize ,and Shape 

```python
#crop an image
#strat form y_start:y_end menas row start row end logics

crop=img_rgb[500:600,200:300]
print(f"Crop Shape: {crop.shape}")

#Resize it to 200x200
resized_crop=cv2.resize(crop,(200,200))

#Display side by side

fig,ax=plt.subplots(1,3,figsize=(15,5))

ax[0].imshow(img_rgb); ax[0].set_title("Original")
ax[1].imshow(crop);  ax[1].set_title("Crop (100x100)")
ax[2].imshow(resized_crop); ax[2].set_title("Resized (200x200)")
plt.axis("off")
plt.tight_layout()
plt.show()
#convert back to BGR and save
resized_bgr=cv2.cvtColor(resized_crop,cv2.COLOR_RGB2BGR)
success=cv2.imwrite("resize_crop.jpg", resized_bgr)
print(f"Saved Successfully :{success}")
```
